The **assignment operator**, denoted by the `=` symbol, is used to assign a value to a variable in Python.

In other words, it tells Python to **store the value on the right-hand side** in the variable named on the left-hand side.


````python
# Examples
x = 5
y = 2
my_variable = 3
x*y + my_variable
````

The following is **not valid Python**:


Assignment is directional: the variable name belongs on the left, so `x = 1` is valid but `1 = x` is not.


<div style="border-left: 5px solid #b00020; padding: 10px; background-color: #fff5f5;">
<strong style="color:#b00020;">Caution: Assignment Is Not Symmetric</strong><br><br>

In mathematics, <span aria-label="x equals 1"> x = 1 </span> and <span aria-label="1 equals x">1 = x </span> mean the same thing logically.

In programming, this is <strong>not</strong> true.  
The variable name must always appear on the <strong>left</strong>, and the value being assigned must appear on the <strong>right</strong>.
</div>

````python
x = 5
x = x + 1
print(x)
````

<div style="border-left: 5px solid #b00020; padding: 10px; background-color: #fff5f5;">
<strong style="color:#b00020;">Caution: Assignment vs. Equality</strong><br><br>

In mathematics, the equation $ x = x + 1 $ has no solution.

In programming, this does <strong>not</strong> mean the same thing.  
When you write

<code>x = x + 1</code>

Python first evaluates the expression on the <strong>right-hand side</strong>, then assigns the result to the variable on the <strong>left-hand side</strong>.
</div>


````python
import builtins

print("This works!")

# A bad variable name can hide a built-in function.
print = 5
builtins.print("After `print = 5`, the name print refers to:", print)

# Remove our accidental variable so Python can find the built-in print again.
del print
print("The built-in print function works again.")
````

## Lists

A **list** stores an ordered, mutable collection of values. Lists may contain different Python types, although data-science tables usually work best when a column has a consistent meaning/type.

````python
my_list = [1, 2, 3, "four"]
print(my_list)
print(my_list[0])
print(my_list[-1])
````

````python
my_list.append(5)
print(my_list)
````

Because lists are mutable, methods such as `append` can change the existing object.

## Tuples

A tuple is an ordered collection like a list, but tuples are **immutable**: their elements cannot be replaced after creation. A one-item tuple needs a trailing comma: `(1,)`.

````python
unknown = (1)
known = (1,)
print(type(unknown), type(known))
````

````python
my_tuple = (1, "two", 3.0, "Cool", "Beans")
print(my_tuple)
````

````python
tuple_a = (1, 2, 3)
try:
    tuple_a[2] = "four"
except TypeError as error:
    print("Expected TypeError:", error)
````

## Sets

A set stores an unordered collection of unique elements. Sets are useful for membership tests, removing duplicates, and set operations such as intersection and union.

````python
my_set = {1, 2, 3, 4, 1, 2, 3, 5}
print(my_set)
````

````python
set_a = {1, 2, 3, "cat", "dog"}
set_b = {0, 2, 4, "hamster", "cat"}
print(set_a.intersection(set_b))
````

## Dictionaries

A dictionary stores **key–value pairs**. Modern Python dictionaries preserve insertion order, but we normally access values by key rather than numerical position. Dictionaries are especially important because JSON/API data often become dictionaries and lists in Python.

````python
school_dict = {}
school_dict['UF'] = 'FL'
school_dict
````

````python
released = {
    "NES": 1983,
    "SNES": 1990,
    "N64": 1996,
}
released["GameCube"] = 2001
print(released)
````

````python
for system, year in released.items():
    print(system, "-->", year)
````
