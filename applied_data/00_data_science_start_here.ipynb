## 5. Use — Meet the DataFrame

A pandas **DataFrame** is a table with labeled rows and columns.

For beginner data science, always ask:

> **What does one row represent?**

In the table below, **one row represents one bird observation**.

A **column** stores one variable, such as species, count, or location. A **row** stores one record.

`import pandas as pd` loads the pandas package and gives it the short name `pd`.

````python
from IPython.display import display
import pandas as pd

birds = pd.DataFrame(
    [
        {"species": "Northern Cardinal", "count": 3, "location": "Campus"},
        {"species": "Blue Jay", "count": 2, "location": "Campus"},
        {"species": "American Crow", "count": 5, "location": "Paynes Prairie"},
        {"species": "Northern Cardinal", "count": 4, "location": "Paynes Prairie"},
    ]
)

display(birds)
````

### Inspect before analyzing

These commands answer different questions:

- `birds.head()` — What do the first rows look like?
- `birds.shape` — How many rows and columns are there?
- `birds.columns` — What are the column names?
- `birds.dtypes` — What type is stored in each column?

`head()` is a **method**, so it uses parentheses. `shape`, `columns`, and `dtypes` are **attributes**, so they do not.

````python
print("shape:", birds.shape)
print("columns:", list(birds.columns))
print("column types:")
print(birds.dtypes)

display(birds.head())
````

### Select and filter

`birds["count"]` selects one column. pandas calls one labeled column a **Series**.

A comparison such as `birds["count"] >= 4` produces `True` or `False` for each row. Putting that condition inside `birds[...]` keeps only the rows where the condition is `True`.

**Predict:** which rows will remain?

````python
display(birds["count"])

at_least_four = birds[birds["count"] >= 4]
display(at_least_four)
````

### Make a basic summary

A summary condenses many values into a few useful numbers.

For a numeric column:

- `.count()` counts known values;
- `.mean()` computes the average of known values;
- `.min()` and `.max()` find the smallest and largest known values;
- `.describe()` gives several common summaries at once.

````python
print("known counts:", birds["count"].count())
print("mean count:", birds["count"].mean())
print("minimum count:", birds["count"].min())
print("maximum count:", birds["count"].max())

display(birds["count"].describe())
````

### Ask one grouped question

Suppose we ask: **How many reported birds are represented at each location?**

`groupby("location")` collects rows with the same location. `["count"]` chooses the column to combine. `.sum()` adds the counts in each group.

**Predict:** what will the two location totals be?

````python
location_totals = birds.groupby("location")["count"].sum()
display(location_totals)
````

You do not need to memorize `groupby()` yet. Focus on the idea: **group rows by a label, then combine values inside each group**.

### Missing values are information too

Real data are often incomplete. A count may be missing because nobody recorded it or because a source did not provide it.

`None` becomes a missing value in a numeric pandas column, often displayed as `NaN`.

**Missing does not mean zero.**

Predict how many values are known below.

````python
counts_with_gap = pd.Series([3, None, 5])

print(counts_with_gap.isna())
print("known values:", counts_with_gap.count())
print("missing values:", counts_with_gap.isna().sum())
print("mean of known values:", counts_with_gap.mean())
````

### Reuse a calculation with a function

A **function** is a named set of instructions. `def` defines the function, a **parameter** receives input, and `return` sends a result back.

Defining a function does not run it.

**Predict:** what will the call return?

````python
def double_count(count):
    return 2 * count

print(double_count(3))
````

### Read an error message

This is an **expected error example**, shown as text so that Run All can still finish:

```python
birds["counts"]
# KeyError: 'counts'
```

The DataFrame has a column named `count`, not `counts`.

When an error appears:

1. Read the last line for the error type.
2. Look at the line of code that failed.
3. Check spelling and cell order.
4. Inspect the object with tools such as `type(...)` or `birds.columns`.

For a `NameError`, check whether the variable was spelled correctly and whether its defining cell ran. For a `TypeError`, inspect the types of the values involved.

## 7. Connect — The beginner data-science workflow

Most MAD 2502 data projects can be described with this loop:

1. **Question** — What do you want to know?
2. **Get data** — file, API, webpage, or another source.
3. **Inspect** — rows, columns, types, and missing values.
4. **Clean** — fix types, names, missing values, duplicates, or structure.
5. **Analyze** — summaries, comparisons, models, or simulations.
6. **Visualize** — make a graph that answers a question.
7. **Interpret** — explain what the result means and what it does not mean.
8. **Revise** — ask a better question and repeat.

You are **not expected to know all of Python before doing data science**. You learn Python by using it to answer increasingly interesting questions.
