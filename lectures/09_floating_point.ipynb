## Understand: binary integers

Binary uses powers of 2. For example,

$$101101_2 = 1(2^5)+0(2^4)+1(2^3)+1(2^2)+0(2^1)+1(2^0)=45.$$

Each bit contributes a power of 2 depending on its position.

```python
print(int('101101', 2))
print(bin(45))
```

## Predict → Run: decimal fractions

Before running, predict whether the comparison is `True` or `False`. Mathematically $0.1+0.2=0.3$, but most decimal fractions do not have a finite binary expansion.

```python
print(0.1 + 0.2)
print((0.1 + 0.2) == 0.3)
```

The issue is not that Python is doing the arithmetic randomly. It is doing arithmetic on nearby representable binary floating-point numbers.

## IEEE-754 binary64

Python's usual `float` is normally an IEEE-754 **binary64** value with 64 bits:

- 1 sign bit $s$
- 11 exponent bits
- 52 stored fraction bits

For a normal finite number, the conceptual form is

$$(-1)^s 2^{e-1023}(1+f),$$

where $e$ is the unsigned exponent field and $f$ is the binary fraction represented by the 52 fraction bits. Special exponent patterns represent zero/subnormal values and infinities/NaNs.

### Normal vs. subnormal

The **smallest positive normal** binary64 value is about $2.225\times10^{-308}$. Smaller positive values can still be represented as **subnormal** numbers, down to about $4.94\times10^{-324}$, but with reduced precision.

```python
import sys

print(sys.float_info.min)       # smallest positive normal float
print(sys.float_info.epsilon)   # spacing near 1.0
print(sys.float_info.max)
```

## Absolute and relative error

If $x$ is the true value and $\hat{x}$ is an approximation,

$$E_{abs}=|x-\hat{x}|,$$

$$E_{rel}=\frac{|x-\hat{x}|}{|x|}\quad (x\ne0).$$

Absolute error has the same units as the quantity. Relative error measures error compared with the size of the true value.

```python
true_value = 1000.0
approximation = 999.5
absolute_error = abs(true_value - approximation)
relative_error = absolute_error / abs(true_value)
print(absolute_error, relative_error)
```

## Use: tolerant comparisons

For computed floats, ask whether values are **close enough for the problem**, not whether their bit patterns are identical. Python's `math.isclose` lets us specify relative and absolute tolerances.

```python
import math
import numpy as np

a = 0.1 + 0.2
b = 0.3
print(math.isclose(a, b, rel_tol=1e-12, abs_tol=1e-12))
print(np.isclose(a, b))
```

## Cancellation

Subtracting nearly equal floating-point numbers can lose significant digits because the leading digits cancel.

```python
x = 1.000000000000001
y = 1.0
print(x - y)
```
