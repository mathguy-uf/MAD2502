## Learning goals

## Understand: three broad kinds of mistakes

- **Syntax error:** Python cannot parse the code.
- **Runtime error / exception:** the code is valid Python, but something goes wrong while it runs.
- **Logical error:** the program runs, but the answer is wrong.

Syntax-error examples are shown as text in this notebook so **Run All** does not stop.

### Expected syntax error

This is intentionally invalid because the colon is missing:

```python
if x > 0
    print(x)
```

When Python reports a syntax error, inspect the indicated line and also the line immediately above it.

## Runtime errors and tracebacks

The final line of a traceback usually gives the exception type and the most useful short message.

````python
examples = [
    ('ZeroDivisionError', lambda: 1 / 0),
    ('IndexError', lambda: [10, 20][5]),
    ('ValueError', lambda: int('cat')),
]

for label, action in examples:
    try:
        action()
    except Exception as error:
        print(label, '->', type(error).__name__ + ':', error)
````

## Predict → Run: catching a specific exception

Predict what happens when the divisor is 0. Catching a **specific** exception is usually better than catching every possible exception.

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

A returned `None` needs interpretation. In a real project you might instead raise an error, report a warning, or mark a value missing depending on what the result means.

## Raise an exception when input is invalid

A function can protect its assumptions with `raise`.

````python
def square_root_input(x):
    if not isinstance(x, (int, float)):
        raise TypeError('x must be numeric')
    if x < 0:
        raise ValueError('x must be nonnegative')
    return x**0.5

print(square_root_input(9))

try:
    square_root_input(-4)
except ValueError as error:
    print('Expected ValueError:', error)
````

## Logical errors

Logical errors do not necessarily produce a traceback. You need evidence that the result disagrees with something you know.

````python
def bad_average(a, b):
    return a + b / 2

print('bad result:', bad_average(10, 20))
print('expected:', 15)
````

The formula should be `(a + b) / 2`. Parentheses are part of the mathematics, not just formatting.

## Use: small tests

Test values for which you already know the answer. `assert` is useful during development because it stops if a condition is false.

````python
def average(a, b):
    return (a + b) / 2

assert average(10, 20) == 15
assert average(-2, 2) == 0
print('tests passed')
````

## A beginner debugging routine

1. Read the **last line** of the traceback.
2. Identify the line in your code where Python stopped.
3. Inspect the values and types involved: `print(value)`, `type(value)`, `len(value)`, `df.shape`, etc.
4. Reduce the problem to the smallest example that still fails.
5. Check assumptions: filename, column name, units, type, missing data, boundary condition.
6. Change one thing at a time.
7. Rerun from a clean kernel if execution order may be part of the problem.
