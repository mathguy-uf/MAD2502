## 3. Run — Lists and dictionaries

Two Python objects appear constantly in data science.

A **list** stores several values in order:

```python
temperatures = [78, 81, 83, 80]
```

A **dictionary** stores labeled values as key/value pairs:

```python
observation = {
    "species": "Northern Cardinal",
    "count": 3,
}
```

APIs often return a **list of dictionaries**.


An **object** is a value Python can work with: a number, text, a list, or a table. Objects have a type. Some have named information (**attributes**) and actions (**methods**). Indexing means selecting an item. List positions start at **0**; dictionary keys are labels such as `"species"`.

````python
observations = [
    {"species": "Northern Cardinal", "count": 3},
    {"species": "Blue Jay", "count": 2},
    {"species": "American Crow", "count": 5},
]

print(observations[0])
print(observations[0]["species"])
````

## 5. Use — Meet the DataFrame

A pandas **DataFrame** is a table with labeled rows and columns.

For beginner data science, the most important question is:

> **What does one row represent?**

In the table below, one row represents one bird observation.


A **package** is a collection of reusable Python tools. `import pandas as pd` makes pandas available under the short name `pd`. `pd.DataFrame(...)` calls a function-like constructor to build a table. A **column** holds one variable (such as count); a **row** holds one record.

````python
from IPython.display import display
import pandas as pd

birds = pd.DataFrame(
    [
        {"species": "Northern Cardinal", "count": 3, "location": "Campus"},
        {"species": "Blue Jay", "count": 2, "location": "Campus"},
        {"species": "American Crow", "count": 5, "location": "Paynes Prairie"},
        {"species": "Northern Cardinal", "count": 4, "location": "Paynes Prairie"},
    ]
)

birds
````

### Inspect before analyzing

These commands answer different questions:

- `birds.head()` — What do the first rows look like?
- `birds.shape` — How many rows and columns?
- `birds.columns` — What are the column names?
- `birds.dtypes` — What type is stored in each column?


`shape`, `columns`, and `dtypes` are attributes: no parentheses. `head()` is a method: parentheses call its action. `display(...)` shows a table in a notebook; `print(...)` prints text.

````python
print("shape:", birds.shape)
print("columns:", list(birds.columns))
print()
print(birds.dtypes)
display(birds.head())
````

### Ask one simple question

Suppose we ask: **How many birds were reported at each location?**


`groupby("location")` collects rows with the same location. `["count"]` chooses the quantity to add. `.sum()` adds the known counts in each group. **Predict:** what will the two location totals be?

````python
location_totals = (
    birds.groupby("location")["count"]
    .sum()
)

location_totals
````

You do not need to memorize `groupby()` yet. Focus on the idea: **group rows by a label, then combine values inside each group**.


## Select one column or a few rows
`birds["count"]` selects a column, called a **Series**. `birds["count"] >= 4` compares each count with 4 and produces True/False values. Putting that condition inside `birds[...]` keeps the True rows. **Predict:** which two rows remain?

````python
display(birds["count"])
display(birds[birds["count"] >= 4])
````

## Reuse a calculation with a function
A **function** is a named set of instructions. `def` defines it; the indented lines are its body. A **parameter** receives an input. `return` sends a result back. Defining a function does not run it. **Predict:** what will the call return?

````python
def double_count(count):
    return 2 * count

print(double_count(3))
````

**Change:** pass 5 instead of 3. Then write a function that adds 1.

## Missing values are information too
Data may be missing because nobody recorded a count, a question was skipped, or a sensor failed. `None` becomes a missing value in a numeric pandas column, often displayed as `NaN`. Missing does not mean zero. **Predict:** how many values are known below?

````python
counts_with_gap = pd.Series([3, None, 5])
print(counts_with_gap.isna())
print("known values:", counts_with_gap.count())
print("mean of known values:", counts_with_gap.mean())
````

## 7. Connect — The beginner data-science workflow

Most MAD 2502 data projects can be described with this loop:

1. **Question** — What do you want to know?
2. **Get data** — file, API, webpage, or other source.
3. **Inspect** — rows, columns, types, missing values.
4. **Clean** — fix types, names, missing values, duplicates, or structure.
5. **Analyze** — summaries, comparisons, models, or simulations.
6. **Visualize** — make a graph that answers a question.
7. **Interpret** — explain what the result means and what it does not mean.
8. **Revise** — ask a better question and repeat.

You are **not expected to know all of Python before doing data science**. You learn Python by using it to answer increasingly interesting questions.

### If you get an error

Do not start over. Read the final lines of the error and ask:

- What line failed?
- What object was Python working with?
- Did I misspell a variable or column name?
- Is the value the type I expected?
- Did I run the cells above this one?
