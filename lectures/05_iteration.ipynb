## Understand: `for` loops

A `for` loop visits each item in a sequence. The loop variable takes one value at a time.

````python
for value in [4, 7, 2]:
    print(value)
````

## Predict → Run: an accumulator

Before running, write the value of `total` after every iteration.

````python
total = 0

for value in [4, 7, 2, 9]:
    total += value
    print('running total:', total)

print('final total:', total)
````

## `range`

`range(start, stop, step)` generates integer values up to, but not including, `stop`.

````python
print(list(range(5)))
print(list(range(2, 10, 2)))
````

## Counters and conditions inside loops

````python
numbers = [3, 8, 12, 1, 7, 10]
count = 0

for number in numbers:
    if number >= 8:
        count += 1

print(count)
````

## `while` loops

A `while` loop repeats while a condition remains true. Ask three questions: **What changes? What stays fixed? Why will it stop?**

````python
n = 5

while n > 0:
    print(n)
    n -= 1

print('done')
````

## A nonterminating-loop warning

The following pattern would never stop because `x` never changes:

```python
x = 1
while x > 0:
    print(x)
```

Do not run an intentionally infinite loop in a shared notebook. Fix the update or add a sensible iteration limit.

## Use: repeated approximation

This simple loop repeatedly averages a guess with `2/guess`. It approaches $\sqrt{2}$.

````python
guess = 1.0

for step in range(6):
    guess = 0.5 * (guess + 2 / guess)
    print(step, guess)
````

## `enumerate` and `zip`

`enumerate` gives an item and its position. `zip` lets us walk through matching sequences together.

````python
species = ['Blue Jay', 'Northern Cardinal', 'American Crow']
counts = [14, 23, 9]

for index, name in enumerate(species):
    print(index, name)

for name, count in zip(species, counts):
    print(name, count)
````

## Optional: list comprehensions

A list comprehension is compact syntax for building a list. Use it only when it remains easy to read.

````python
squares = [number**2 for number in range(1, 6)]
print(squares)
````
