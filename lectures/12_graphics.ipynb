## Understand: Figure and Axes

Matplotlib has a useful hierarchy:

```text
Figure
└── Axes
    ├── plot / scatter / bar / hist
    ├── title
    ├── x-label and y-label
    ├── legend
    └── annotations
```

The **Figure** is the whole canvas. An **Axes** is the plotting area where data and labels are drawn. Most of our work will use `fig, ax = plt.subplots()`.

## Predict → Run: your first plot

Before running, list the four coordinates that will be plotted and predict the general shape.

````python
x = np.array([0, 1, 2, 3])
y = x**2

fig, ax = plt.subplots()
ax.plot(x, y, marker='o')
ax.set(xlabel='x', ylabel='y', title='A quadratic pattern')
ax.grid(True, alpha=0.3)
plt.show()
````

## Plot layers and style strings

Each method call adds or changes something on the Axes. A compact style string such as `'ko--'` means **black (`k`) circles (`o`) with a dashed line (`--`)**. You may also use named keyword arguments, which are often easier to read.

````python
x = np.linspace(0, 2 * np.pi, 200)

fig, ax = plt.subplots()
ax.plot(x, np.sin(x), 'k-', label='sin(x)')
ax.plot(x, np.cos(x), marker='o', markevery=20, linestyle='--', label='cos(x)')
ax.axhline(0, linewidth=1)
ax.set(xlabel='x', ylabel='value', title='Two functions on one Axes')
ax.legend()
plt.show()
````

## Scatter plots

Use a scatter plot when individual paired observations matter and a connecting line would imply a relationship that is not actually present.

````python
rng = np.random.default_rng(2502)
heights = rng.normal(170, 8, 30)
weights = 0.45 * heights - 8 + rng.normal(0, 5, 30)

fig, ax = plt.subplots()
ax.scatter(heights, weights, marker='o')
ax.set(xlabel='Height (cm)', ylabel='Example weight measure', title='Example paired observations')
plt.show()
````

## Bar charts and histograms

A **bar chart** compares categories. A **histogram** groups numerical values into intervals to show a distribution.

````python
species = ['Blue Jay', 'Cardinal', 'Crow']
counts = [24, 37, 18]
fig, ax = plt.subplots()
ax.bar(species, counts)
ax.set(ylabel='Count', title='Example categorical counts')
plt.show()

fig, ax = plt.subplots()
ax.hist(rng.normal(size=500), bins=20, edgecolor='black')
ax.set(xlabel='Value', ylabel='Frequency', title='Example distribution')
plt.show()
````

## 3-D graphics

A 3-D Axes can display a curve or surface with three coordinates. Do not use 3-D merely because it looks impressive; use it when a third dimension represents meaningful information.

````python
t = np.linspace(0, 6 * np.pi, 300)
x = np.sin(t)
y = np.cos(t)
z = t

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(x, y, z)
ax.set(xlabel='x', ylabel='y', zlabel='z', title='A helix')
plt.show()
````

## Maps: optional extension

Map plotting adds a geographic coordinate system and map projection. The textbook introduces maps after ordinary 2-D/3-D plotting. A common Python package is **Cartopy**, but it is heavier than the packages required for the core course and may not be installed in every environment.

The important beginner idea is: longitude/latitude data must be interpreted in an appropriate geographic coordinate system. For capstone work, use a map only when geography helps answer the question.

## Animation: the basic idea

An animation repeatedly updates artists on a Figure. The mathematics is still ordinary data: at frame `i`, decide what should be visible.

The example below animates a point moving along a sine curve. It uses an HTML player so it can display in many notebook environments without requiring a desktop GUI.

````python
x = np.linspace(0, 2 * np.pi, 60)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y, linewidth=1)
point, = ax.plot([], [], 'ko')
ax.set_xlim(x.min(), x.max())
ax.set_ylim(-1.2, 1.2)
ax.set_title('A point moving along sin(x)')

def update(frame):
    point.set_data([x[frame]], [y[frame]])
    return point,

animation = FuncAnimation(fig, update, frames=len(x), interval=60, blit=True)
plt.close(fig)
HTML(animation.to_jshtml())
````

## Use: a small Sierpiński example

Repeatedly move halfway from the current point toward a randomly chosen triangle vertex. A simple rule produces a surprising fractal structure.

````python
rng = np.random.default_rng(2502)
vertices = np.array([[0.0, 0.0], [1.0, 0.0], [0.5, np.sqrt(3) / 2]])
point = np.array([0.2, 0.1])
points = []

for _ in range(5000):
    vertex = vertices[rng.integers(0, 3)]
    point = (point + vertex) / 2
    points.append(point.copy())

points = np.array(points)
fig, ax = plt.subplots()
ax.scatter(points[:, 0], points[:, 1], s=1)
ax.set_aspect('equal')
ax.set_title('Sierpiński triangle by iteration')
ax.axis('off')
plt.show()
````
