---



### Boolean Logic and History
**Historical note:** Boolean logic is named after mathematician George Boole.


The Boolean type (`bool`) is named after George Boole (1815–1864).

Boole, a largely self-taught mathematician, showed that logical reasoning could be expressed using algebra. About a century later, Claude Shannon demonstrated that Boole’s 0/1 logic is exactly how electrical circuits operate.

As a result, every modern computer ultimately relies on Boolean logic—and Python’s `True` and `False` values are direct descendants of Boole’s ideas.

### Floating-Point Comparisons

One subtle issue in computational mathematics is that floating-point numbers are stored using finite precision.

As a result, expressions that are mathematically equal may not be *exactly* equal in a computer.

For example:


````python
0.1 + 0.2 == 0.3
````

The result may be `False` because the computer stores approximations of these decimal numbers.

Instead of checking exact equality, we often compare the difference to a small tolerance:

```python
abs(a - b) < tolerance
```

For example:


````python
abs(0.1 + 0.2 - 0.3) < 0.001
````

## Logical Operators

**Logical operators** combine logical expressions, which we will call $P$ and $Q$.  
In Python, the fundamental logical operators are **`and`**, **`or`**, and **`not`**.

| Operator | Example     | Meaning |
|---------:|-------------|---------|
| `and` | `P and Q` | `True` if **both** $P$ and $Q$ are `True`; `False` otherwise |
| `or`  | `P or Q`  | `True` if **either** $P$ or $Q$ is `True` (inclusive or); `False` otherwise |
| `not` | `not P`   | `True` if $P$ is `False`; `False` otherwise |

These operators are commonly used to build compound logical conditions in Python.

### Errors and Exceptions in Python

When Python encounters something it cannot do, it raises an **error** (also called an **exception**).

For example, Python will raise a **`ZeroDivisionError`** when you attempt to divide by zero.

**Expected error:** the next example intentionally divides by zero so you can see what a Python traceback looks like. Read the final line of the error message before continuing.

Try the following in a **code cell**:

```
1 / 0
```

You will see output similar to:

```
Traceback (most recent call last):
  File "<stdin>", line 1, in <module>
ZeroDivisionError: division by zero
```

This message is called a **traceback**. It tells you:
- What kind of error occurred
- Where the error happened
- Why Python stopped executing the code

---

**Important:**  
Errors are a normal and expected part of programming. Reading error messages is a skill, and it gets easier with practice.

We will learn how to **handle exceptions** later in the course. For now, you should deliberately try simple errors like this so you become comfortable recognizing them when they appear.


````python
try:
    1 / 0
except ZeroDivisionError as error:
    print("Expected ZeroDivisionError:", error)
````
