## Understand — lists, tuples, and dictionaries
A list is an ordered, changeable collection. A tuple is ordered and cannot have its items reassigned. A dictionary associates keys with values. Index positions start at zero. These are the building blocks of JSON and DataFrames.

```python
def find_divisors(n1, n2):
    """Assumes that n1 and n2 are positive integers.
    Returns a tuple containing all common divisors of n1 and n2"""
    divisors = ()  # empty tuple
    for i in range(1, min(n1, n2) + 1):
        if n1 % i == 0 and n2 % i == 0:
            divisors = divisors + (i,)
    return divisors


divisors = find_divisors(800, 2000)
print(divisors)
total = 0
for d in divisors:
    total += d
print("The sum of divisors is:", total)
```

```python
def find_extreme_divisors(n1, n2):
    """Assumes that n1 and n2 are positive integers.  Returns
    a tuple containing the smallest common divisor>1 and the
    largest common divisor of n1 and n2"""
    divisors = ()  # the empty tuple
    min_value, max_value = None, None
    for i in range(2, min(n1, n2) + 1):
        if n1 % i == 0 and n2 % i == 0:
            if min_value is None or i < min_value:
                min_value = i
            if max_value is None or i > max_value:
                max_value = i
    return (min_value, max_value)
```

```python
min_divisor, max_divisor = find_extreme_divisors(100, 200)
print(min_divisor, max_divisor)
```

## Equality versus identity
Equal contents do not require the same object. The nested list below shares the `florida` list. Predict what changes when that list is appended to.

```python
ivys = ["Harvard", "Yale"]
florida = ["UF", "UCF"]
jux = [ivys, florida]  # juxtaposition
tog = [["Harvard", "Yale"], ["UF", "UCF"]]
print(jux)
print(tog)
print(jux == tog)
```

```python
print(jux == tog)
print(id(jux) == id(tog))
print("Id of tog=", id(tog))
print("Id of jux=", id(jux))
```

```python
florida.append("FSU")  # append to list
print(jux)
print(tog)
```

```python
for i in tog:  # each item in the list
    print("schools: ", i)
    print(" which contains")  # 1 space in front
    for u in i:  # each item in the sublist
        print("  ", u)  # 2 spaces in front
```

```python
t1 = (1, "two", 3)
t2 = ("apple", "onion")
t3 = t1 + t2  # created a tuple with t1
print("t1 before append: ", t1)
print("t3 before append to t1:", t3)
t1 = t1 + ("5",)  # Create a new tuple and rebind t1; tuples do not mutate.
print("t1 after append: ", t1)
print("t3 after append to t1:", t3)  # check if t3 is affected
```

```python
first_list = [1, 2, 3]
second_list = [4, 5, 6]
combined_list = first_list + second_list  # creates a new list
print("combined_list=", combined_list)
first_list.extend(second_list)  # mutated first_list
print("first_list=", first_list)
first_list.append(second_list)  # mutated first_list again!
print("first_list=", first_list)
print("combined_list=", combined_list)  # combined_list should stay the same
```

## Intentional logic bug
The next function mutates the list it is iterating over. Predict which shared value it accidentally skips. The following cell shows the repair: iterate over a copy. The faulty example is intentionally preserved.

```python
def remove_duplicates(first_list, second_list):
    """Assumes that first_list and second_list are lists
    Removes any element from first_list that also occurs in second_list"""
    for e1 in first_list:
        if e1 in second_list:
            first_list.remove(e1)


first_list = [1, 2, 3, 4]
second_list = [1, 2, 5, 6]
remove_duplicates(first_list, second_list)
print("first_list=", first_list)
```

```python
def remove_duplicates(first_list, second_list):
    """Assumes that first_list and second_list are lists
    Removes any element from first_list that also occurs in second_list"""
    for e1 in first_list[:]:  # Wow! So subtle! What did that do?
        if e1 in second_list:
            first_list.remove(e1)


first_list = [1, 2, 3, 4]
second_list = [1, 2, 5, 6]
remove_duplicates(first_list, second_list)
print("first_list=", first_list)
```

## Dictionaries and iteration
