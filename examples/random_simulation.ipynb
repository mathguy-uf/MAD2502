## Understand — repeated trials
A fair die gives each integer 1 through 6 the same chance. A fair coin has probability 0.5 of heads. A simulation repeats those assumptions in code.

**Predict:** should a single set of ten flips have exactly five heads?

```python
import random
import numpy as np
import matplotlib.pyplot as plt

rng = random.Random(2502)
rolls = [rng.randint(1, 6) for _ in range(10)]
heads = 0
for _ in range(10):
    if rng.random() < 0.5:
        heads += 1
print("Ten die rolls:", rolls)
print("Fraction of heads:", heads / 10)
```

## Use — compare many runs
NumPy can simulate many independent binomial counts efficiently. A **trial** here is a whole set of coin flips. Compare how much the fraction of heads varies across trials. The sample standard deviation of simulated fractions estimates their spread; the standard error of their mean is smaller by $\sqrt{n_{trials}}$. They answer different questions.

```python
generator = np.random.default_rng(2502)
number_of_trials = 2000
fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharex=True, sharey=True)
for ax, flips_per_trial in zip(axes, [10, 100]):
    fractions = (
        generator.binomial(flips_per_trial, 0.5, size=number_of_trials)
        / flips_per_trial
    )
    ax.hist(fractions, bins=np.linspace(0, 1, 21), edgecolor="black")
    ax.set(
        xlabel="Fraction heads",
        ylabel="Trials",
        title=f"{flips_per_trial} flips per trial",
    )
    print(
        flips_per_trial, "flips; mean:", fractions.mean(), "SD:", fractions.std(ddof=1)
    )
plt.tight_layout()
plt.show()
```

## Compare uncertainty in a simulated mean
The following approximate interval is **mean ± 1.96 × standard error of the mean**, using independent simulation trials. It describes uncertainty in the estimated model mean, not where 95% of individual trials lie. The approximation is appropriate here because we average many independent bounded trials.

```python
fractions = generator.binomial(100, 0.5, size=number_of_trials) / 100
mean = fractions.mean()
standard_error = fractions.std(ddof=1) / np.sqrt(number_of_trials)
print(
    "Approximate interval for model mean:",
    mean - 1.96 * standard_error,
    mean + 1.96 * standard_error,
)
```
