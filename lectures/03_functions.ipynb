## Understand: define and call

A function definition begins with `def`. Parameters are names in the definition; arguments are the actual values supplied when the function is called.

```python
def my_max(a, b):
    if a > b:
        return a
    return b

print(my_max(4, 9))
```

## Predict → Run: `return` is not `print`

Predict what is stored in `answer`. A function can return a value without printing it. Returned values can be reused in later calculations.

```python
def square(x):
    return x**2

answer = square(5)
print(answer)
print(square(3) + square(4))
```

A function with no explicit `return` returns the special value `None`.

```python
def greeting(name):
    print(f'Hello, {name}!')

result = greeting('Gator')
print('returned value:', result)
```

## Positional, keyword, and default arguments

Positional arguments are matched by order. Keyword arguments are matched by name. A default value makes an argument optional.

```python
def distance(x1, y1, x2=0, y2=0):
    return ((x2 - x1)**2 + (y2 - y1)**2)**0.5

print(distance(3, 4))
print(distance(x1=1, y1=1, x2=4, y2=5))
```

## Scope

Variables created inside a function are normally **local** to that function. A local variable can have the same name as a variable outside without replacing it.

```python
x = 100

def demo_scope():
    x = 5
    print('inside:', x)

demo_scope()
print('outside:', x)
```

## Functions can call functions

Breaking a problem into small functions often makes code easier to test and explain.

```python
def celsius_to_fahrenheit(celsius):
    return (9 / 5) * celsius + 32

def describe_temperature(celsius):
    fahrenheit = celsius_to_fahrenheit(celsius)
    return f'{celsius} C is {fahrenheit:.1f} F'

print(describe_temperature(25))
```

## Optional preview: functions as objects

Python can pass a function to another function. You will see this idea again in numerical methods.

```python
def apply_twice(function, value):
    return function(function(value))

print(apply_twice(square, 2))
```
