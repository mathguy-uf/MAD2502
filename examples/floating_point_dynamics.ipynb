## Understand and predict
The doubling map is $D(x)=2x\bmod1$. In exact arithmetic, $1/9$ follows a six-point cycle. Binary floating-point cannot represent $1/9$ exactly. Predict whether the computed orbit will keep returning to its initial value.

```python
from fractions import Fraction

import matplotlib.pyplot as plt

exact = Fraction(1, 9)
approximate = 1 / 9
exact_orbit, float_orbit = [], []
for _ in range(65):
    exact_orbit.append(float(exact))
    float_orbit.append(approximate)
    exact = (2 * exact) % 1
    approximate = (2 * approximate) % 1
fig, ax = plt.subplots()
ax.plot(exact_orbit, "o-", markersize=3, label="Exact rational arithmetic")
ax.plot(float_orbit, "x--", markersize=4, label="Binary floating-point")
ax.set(
    xlabel="Iteration", ylabel="Value", title="Doubling amplifies representation error"
)
ax.legend()
plt.show()
```

## Use — another iterative map
Repeat $x\mapsto x^2-2$ starting from 0.1. This is a deterministic orbit, not random data. Its histogram describes these iterates; it does not establish a probability model for real observations.

```python
value = 0.1
orbit = []
for _ in range(2000):
    orbit.append(value)
    value = value**2 - 2
fig, ax = plt.subplots()
ax.hist(orbit, bins=40, edgecolor="black")
ax.set(
    xlabel="Orbit value",
    ylabel="Frequency",
    title="Iterates of x² − 2, starting at 0.1",
)
plt.show()
```
