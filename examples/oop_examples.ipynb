## 1. Understand: your first class

A **class** describes what information an object stores and what actions it can perform.

An **object** is one actual instance of that class.


````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hi! My name is {self.name}."
````

## 2. Use: a `Student` class and `__str__`

Python does not automatically know how you want an object displayed. The special method `__str__` lets you define a readable string representation.


````python
class Student:
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name
        self.scores = []

    def add_score(self, score):
        self.scores.append(score)

    def average(self):
        if not self.scores:
            return None
        return sum(self.scores) / len(self.scores)

    def __str__(self):
        return f"{self.name} ({self.student_id})"
````

````python
student = Student("123", "Karen")
student.add_score(80)
student.add_score(95)

print(student)
print("average:", student.average())
````

## 3. Inheritance: game enemies

Inheritance is useful when several classes share common behavior. A child class can reuse the code in a parent class and then specialize it.


````python
class Enemy:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def take_damage(self, amount):
        self.health = max(0, self.health - amount)

    def __str__(self):
        return f"{self.name}: {self.health} HP"


class Goomba(Enemy):
    def __init__(self):
        super().__init__("Goomba", 10)


class Koopa(Enemy):
    def __init__(self):
        super().__init__("Koopa", 20)
````

### Predict

Both child classes inherit `take_damage`. What will the health values be after each enemy takes 7 damage?


````python
goomba = Goomba()
koopa = Koopa()

goomba.take_damage(7)
koopa.take_damage(7)

print(goomba)
print(koopa)
````

## 4. Inheritance: rectangle and square

This example shows the same idea with mathematics rather than game objects. A square is a special rectangle whose width and height are equal.


````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)
````

````python
rectangle = Rectangle(4, 5)
square = Square(3)

print("rectangle area:", rectangle.area())
print("square area:", square.area())
print("square perimeter:", square.perimeter())
````

## 5. Encapsulation: a treasure chest

Sometimes an object should control how its internal data are accessed.

A name beginning with two underscores is **name-mangled** by Python. This is not absolute security, but it discourages accidental direct access.


````python
class TreasureChest:
    def __init__(self, contents):
        self.__contents = contents

    def open(self):
        return self.__contents


chest = TreasureChest("100 gold coins")
print(chest.open())
````

### Expected error

The following direct access is intentionally shown as something that should fail:

```python
chest.__contents
```

The important lesson is not that Python makes the value impossible to reach; it is that the class provides an intended interface (`open`) for interacting with its state.


## 6. Polymorphism and operator overloading

The `+` operator already behaves differently for numbers, strings, and lists. Classes can define their own meaning for `+` with `__add__`.


````python
class Coin:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        if isinstance(other, Coin):
            return self.value + other.value
        return NotImplemented


coin_1 = Coin(5)
coin_2 = Coin(10)

print(coin_1 + coin_2)
````
