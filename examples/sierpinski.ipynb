## Understand and predict
The chaos game repeatedly chooses a triangle vertex and moves halfway toward it. The rule is simple, but many repetitions reveal a pattern. Use the equilateral triangle with vertices $(0,0)$, $(1,0)$, and $(1/2,\sqrt{3}/2)$.

**Predict:** where does the point $(0,0)$ move when the top vertex is chosen?

```python
import random
import math
import matplotlib.pyplot as plt

rng = random.Random(2502)
vertices = [(0, 0), (1, 0), (0.5, math.sqrt(3) / 2)]
x, y = 0.0, 0.0
x_values, y_values = [], []
for _ in range(5000):
    vertex_x, vertex_y = rng.choice(vertices)
    x = (x + vertex_x) / 2
    y = (y + vertex_y) / 2
    x_values.append(x)
    y_values.append(y)
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(x_values, y_values, s=1)
ax.set(
    xlabel="x",
    ylabel="y",
    title="Sierpiński triangle: halfway to a random vertex",
    aspect="equal",
)
plt.show()
```
