```python
class Cat():
    """A simple attempt to model a cat"""
    
    def __init__(self, name, age):
        """Initialize name and age attributes"""
        self.name = name
        self.age = age    # in months
        
    def sleep(self):
        """Simulate a cat sleeping"""
        print(self.name.title() + ' is now sleeping.')
        
    def destroy(self):
        """simulate a cat eating"""
        print(self.name.title() + ' is now destroying my home.')
       
    def __str__(self):
        """Returns a string representation of self"""
        return 'The cat ' + self.name + ' is ' + str(self.age) + ' months old.'
        
```

## Car state
The odometer should not decrease. Predict what happens when an attempted update is smaller than the current mileage.

```python
class Car():
    """A simple attempt to represent a car"""
    
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer = 0
        
    def __str__(self):
        """Returns a string representing self."""
        return str(self.year) + ' ' + self.make + ' ' + self.model
    
    def update_odometer(self, mileage):
        """set the odometer reading to the given value
        Reject the change if it attempts to roll the odometer back."""
        if mileage >= self.odometer: 
            self.odometer = mileage
        else:
            print("You can not roll back an odometer!")
    
    def increment_odometer(self, mileage):
        """Add the given amount to the odometer reading"""
        if mileage < 0:
            raise ValueError("Mileage increment must be nonnegative.")
        self.odometer += mileage
    
    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print('This car has ' + str(self.odometer) + ' miles on it.' )
```

```python
first_car = Car('Pontiac', 'Sunbird', 1988) #My first car
print(first_car)
first_car.read_odometer()
```

```python
first_car.update_odometer(20000)
```

```python
first_car.read_odometer()
```

```python
first_car.update_odometer(50)
first_car.increment_odometer(50)
first_car.read_odometer()
```

```python
another_car = Car('Kia', 'Optima', 2015)  #A fictional example car
```

```python
for mile in range(100):
    another_car.increment_odometer(mile)
    
another_car.read_odometer()
```

## A mathematical object
A parabola stores coefficients and supplies evaluation, vertex, and plotting methods. The quadratic coefficient must be nonzero.

```python
import numpy as np
import matplotlib.pyplot as plt

class Parabola():

    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    
    def f(self, x):
        return self.a * x**2 + self.b*x +self.c
    
    def __str__(self):
        return str(self.a) +'x^2 + ' + str(self.b) + 'x + ' + str(self.c)
    
    def vertex_x(self):
        if self.a == 0:
            raise ValueError('A parabola requires a nonzero quadratic coefficient.')
        else:
            return -1*self.b /(2.0 * self.a)
        
    def vertex_y(self):
        if self.a == 0:
            raise ValueError('A parabola requires a nonzero quadratic coefficient.')
        else:
            return self.f( self.vertex_x() )
        
    def print_vertex(self):
        print('The vertex is ( ' + str(self.vertex_x()) + ' , ' + str(self.vertex_y()) + ' )')
        
    def graph(self):
        listx = np.arange(-5., 5., 0.2) + self.vertex_x() #puts the vertex in the middle
        plt.plot(listx, self.f( listx ), 'r-')
        plt.xlabel('x-values')
        plt.ylabel('y-values')
        plt.title(str(self.a) +  r'$x^2 +$'  +  str(self.b) + r'$x + $' + str(self.c))
        plt.plot(self.vertex_x(),self.vertex_y(),'ro',label = 'Vertex') 
        if self.a > 0:
            plt.annotate('Minimum', xy=(self.vertex_x(), self.vertex_y()), xytext=(self.vertex_x()+5, self.vertex_y()-2),arrowprops=dict(facecolor='black', shrink=0.05),)
        elif self.a < 0:
            plt.annotate('Maximum', xy=(self.vertex_x(), self.vertex_y()), xytext=(self.vertex_x()-5, self.vertex_y()-2),arrowprops=dict(facecolor='black', shrink=0.05),)
        plt.legend()
        plt.show()
        
```

```python
func = Parabola(-1,2,3)
print(func)
```

```python
func.f(1)
```

```python
func.print_vertex()
```

```python
func.vertex_y()
```

```python
func.graph()
```
