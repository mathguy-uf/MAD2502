A recursive function solves a problem using a smaller version of the same problem.

**Learning rhythm:** Understand → Predict → Run → Change → Use → Practice → Connect.

## Learning goals

You should be able to identify a base case and recursive step, trace a small recursive call by hand, explain why a recursive function terminates, and compare recursion with iteration for simple problems.

## Understand: the two required pieces

A useful recursive function needs:

1. a **base case** that stops the recursion;
2. a **recursive step** that moves the problem closer to the base case.

```python
def factorial(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n must be a nonnegative integer")
    if n == 0:
        return 1
    return n * factorial(n - 1)


print(factorial(5))
```

## Predict → Run: trace `factorial(4)`

Before running, expand it on paper:

`factorial(4)` → `4 * factorial(3)` → ... → `4 * 3 * 2 * 1 * factorial(0)`

The base case returns 1, and then the pending multiplications return in reverse order.

```python
for n in range(6):
    print(n, factorial(n))
```

## Change

Change the base case to `n == 1`. What happens to `factorial(0)`? Then restore the correct version. Small changes to the base case can change whether a recursive function works at all.

## What happens without progress?

If the recursive call does not move toward the base case, Python eventually raises `RecursionError`. We catch it here so the notebook continues.

```python
def broken_countdown(n):
    if n == 0:
        return 0
    return broken_countdown(n + 1)


try:
    broken_countdown(2)
except RecursionError as error:
    print("Expected RecursionError:", error)
```

## Use: Fibonacci numbers

The mathematical definition is recursive, so it translates naturally to code. The direct version is easy to read but repeats a great deal of work for large `n`.

```python
def fibonacci(n):
    if n < 0:
        raise ValueError("n must be nonnegative")
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


print([fibonacci(n) for n in range(10)])
```

For large Fibonacci calculations, an iterative method or memoization is usually more efficient. Recursion is not automatically better; it is a way of expressing a problem.

## Use: palindrome test

A string is a palindrome if its first and last characters match and the smaller string between them is also a palindrome.

```python
def is_palindrome(text):
    cleaned = "".join(character.lower() for character in text if character.isalnum())
    if len(cleaned) <= 1:
        return True
    if cleaned[0] != cleaned[-1]:
        return False
    return is_palindrome(cleaned[1:-1])


print(is_palindrome("racecar"))
print(is_palindrome("A man, a plan, a canal: Panama"))
print(is_palindrome("python"))
```

## Recursion vs. iteration

Both can repeat a process. Recursion is especially natural when the mathematical definition or data structure is recursive. Iteration is often more direct and memory-efficient for straightforward repeated calculations.
