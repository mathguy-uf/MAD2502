## Understand
A random walk adds successive random steps. Brownian motion is a continuous-time model; this notebook draws a **discrete approximation** using independent normal increments with mean 0 and standard deviation $\sqrt{\Delta t}$. We are simulating a mathematical model, not measuring a physical particle.

First examine a simpler walk. **Predict:** what is its final position?

```python
steps = [1, -1, 1, 1, -1]
positions = [0]
for step in steps:
    positions.append(positions[-1] + step)
print(positions)
```

## Use — approximate a Brownian path
A normal distribution is a symmetric bell-shaped probability model. `rng.normal` samples it; `cumsum` forms running totals. `dt` is the time between samples. The starting point is 0.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2502)
number_of_steps = 500
dt = 1 / number_of_steps
increments = rng.normal(0, np.sqrt(dt), number_of_steps)
path = np.concatenate(([0], np.cumsum(increments)))
time = np.linspace(0, 1, number_of_steps + 1)
fig, ax = plt.subplots()
ax.plot(time, path, label="Simulated path")
ax.set(xlabel="Time", ylabel="Position", title="Discrete Brownian approximation")
plt.show()
print("Final displacement:", path[-1])
```

## Compare endpoints
**Predict:** across many simulated paths, should endpoint averages be near zero? Each row below is a new path. We add the actual increments, not merely their signs.

```python
increments = rng.normal(0, np.sqrt(dt), size=(1000, number_of_steps))
endpoints = increments.sum(axis=1)
fig, ax = plt.subplots()
ax.hist(endpoints, bins=30, edgecolor="black")
ax.set(
    xlabel="Endpoint displacement at time 1",
    ylabel="Number of paths",
    title="1,000 simulated endpoints",
)
plt.show()
print("Mean endpoint:", endpoints.mean())
```
