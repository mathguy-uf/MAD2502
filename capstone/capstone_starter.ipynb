## Step 5 — Get one small Python result early

A first result can be a count, percentage, mean, median, grouped summary, ranking, or simple plot.

Predict the output before running the next cell.


````python
counts = data["group"].value_counts(dropna=False)
print("Counts by group:")
print(counts)

print()
print("Mean value by group:")
means = data.groupby("group")["value"].mean()
print(means)
````

### Example: grouped summary

This combines row counts, known-value counts, mean, and median so the denominator is visible.


````python
group_summary = (
    data.groupby("group", as_index=False)
    .agg(
        number_of_rows=("value", "size"),
        known_values=("value", "count"),
        mean_value=("value", "mean"),
        median_value=("value", "median"),
    )
)

group_summary
````

### Example: one useful graph

A graph should answer a question. Label axes and use a meaningful title.


````python
ax = group_summary.plot(
    x="group",
    y="mean_value",
    kind="bar",
    legend=False,
)

ax.set(
    xlabel="Group",
    ylabel="Mean value",
    title="Mean value by group",
)

plt.tight_layout()
plt.show()
````
