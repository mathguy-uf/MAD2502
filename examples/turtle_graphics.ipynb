## 1. Understand

Python's `turtle` module controls a cursor that can move and turn.

A few important commands are:

- `forward(distance)` — move forward;
- `right(angle)` — turn clockwise;
- `left(angle)` — turn counterclockwise;
- `penup()` / `pendown()` — move without drawing / resume drawing.

Instead of `from turtle import *`, this notebook uses `import turtle`. That keeps the source of each command clear.


```python
import turtle

screen = turtle.Screen()
screen.title("MAD 2502 Turtle Graphics")

pen = turtle.Turtle()
pen.shape("turtle")
pen.speed(0)
```

## 2. Predict

Before running the next cell, sketch what you expect to see.

How many line segments are drawn? Where should the turtle finish?


```python
pen.forward(100)
pen.right(90)
pen.forward(100)
pen.right(90)
pen.forward(100)
pen.right(90)
pen.forward(100)
```

## 3. Run

The four repeated moves make a square. Repetition is a clue that a loop would make the code clearer.


```python
pen.clear()
pen.home()

for _ in range(4):
    pen.forward(100)
    pen.right(90)
```

## 5. Use

Package the repeated drawing into a function. Then use a loop to rotate many squares.


```python
def square(t, side_length):
    """Draw a square using turtle t."""
    for _ in range(4):
        t.forward(side_length)
        t.right(90)
```

```python
pen.clear()
pen.home()

for angle in range(0, 360, 10):
    square(pen, 80)
    pen.right(10)
```

The picture is produced by combining two ideas:

- a **function** that draws one square;
- a **loop** that repeats the square after a small rotation.

This is a useful computational pattern: build a small reusable piece, then repeat or transform it.

```python
def polygon(t, side_length, sides):
    """Draw a regular polygon."""
    turn_angle = 360 / sides

    for _ in range(sides):
        t.forward(side_length)
        t.right(turn_angle)
```
