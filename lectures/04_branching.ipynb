Branching lets a program choose what to do based on a condition.

**Learning rhythm:** Understand → Predict → Run → Change → Use → Practice → Connect.

## Learning goals

You should be able to build Boolean conditions, use `if`, `elif`, and `else`, combine conditions with `and`/`or`/`not`, recognize boundary cases, and use branching to validate input or select a computational action.

## Understand: a Boolean condition

A branch begins with an expression that evaluates to `True` or `False`. Python executes an indented block only when its condition is true.

```python
temperature = 88

if temperature > 85:
    print("It is hot.")
```

## Predict → Run: separate `if` statements vs. an `if/elif/else` chain

Before running, predict how many lines print.

```python
x = 7

if x > 5:
    print("A")
if x > 3:
    print("B")
else:
    print("C")
```

```python
x = 7

if x > 5:
    print("A")
elif x > 3:
    print("B")
else:
    print("C")
```

In an `if/elif/else` chain, Python takes the **first true branch** and skips the rest. Separate `if` statements are checked independently.

## Change: test the boundaries

Change `x` to `5`, `3`, and `-1`. Boundary values are where many branching mistakes occur.

## Compound conditions

Use `and` when both conditions must be true, `or` when at least one may be true, and `not` to reverse a Boolean value.

```python
score = 84
submitted = True

if score >= 70 and submitted:
    print("Requirements met.")
```

## Use: interval membership

Mathematical intervals translate naturally into compound conditions. Python also supports chained comparisons.

```python
x = 0.35

if 0 <= x <= 1:
    print("x is in [0, 1]")
else:
    print("x is outside [0, 1]")
```

## Floating-point decisions

Exact equality can be unsafe for computed floating-point values. When the mathematical idea is 'close enough,' use a tolerance.

```python
a = 0.1 + 0.2
b = 0.3
tolerance = 1e-12

if abs(a - b) < tolerance:
    print("a and b are close enough for this calculation")
```

## Use: checking types before arithmetic

For a beginner function, it can be useful to reject unexpected input explicitly.

```python
def my_adder(a, b, c):
    values = (a, b, c)
    if all(isinstance(value, (int, float)) for value in values):
        return a + b + c
    raise TypeError("all three arguments must be numbers")


print(my_adder(1, 2, 3))

try:
    my_adder(1, "2", 3)
except TypeError as error:
    print("Expected TypeError:", error)
```
