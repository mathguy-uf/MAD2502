## Understand: arrays

A Python list can hold many kinds of objects. A NumPy `ndarray` is designed for numerical data and normally stores one data type efficiently.

A 1-D array is like a numerical list. A 2-D array is like a rectangular table of numbers.

```python
import numpy as np

a = np.array([2, 4, 6, 8])
b = np.array([[1, 2, 3], [4, 5, 6]])

print(a)
print(b)
print('shape:', b.shape)
print('dtype:', b.dtype)
```

## Predict → Run: vectorization

Before running the next cell, predict the result of `2 * a`. NumPy applies the operation to every element. This is called **vectorization**.

```python
print(2 * a)
print(a + 10)
print(a**2)
```

## Indexing and slicing

Indexes start at 0. For a 2-D array, use `[row, column]`. A slice selects a range. NumPy slices are usually **views** of the original array, so changing a slice can change the original; use `.copy()` when you need an independent copy.

```python
m = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print(m[0, 1])       # row 0, column 1
print(m[:2, 1:])     # first two rows, last two columns

part = m[:2, :2].copy()
part[0, 0] = -1
print('original still begins with', m[0, 0])
```

## Boolean indexing

A comparison such as `x > 5` produces an array of `True`/`False` values. We can use that Boolean array as a filter.

```python
x = np.array([1, 5, 9, 2, 12, 4])
mask = x > 5
print(mask)
print(x[mask])

# Multiple array conditions use & and |, with parentheses around each comparison.
print(x[(x > 3) & (x < 10)])
```

## Use: basic summaries

NumPy can summarize an entire numerical array. These same ideas appear later in pandas.

```python
data = np.array([4, 7, 3, 10, 6])
print('mean:', data.mean())
print('median:', np.median(data))
print('min:', data.min())
print('max:', data.max())
print('sum:', data.sum())
```

## Elementwise vs. matrix multiplication

For arrays of matching shape, `A * B` multiplies element by element. The `@` operator performs matrix multiplication. These are different operations.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[3, 4], [5, 6]])
print('elementwise:
', A * B)
print('matrix product:
', A @ B)
```
