## Understand
Monte Carlo estimation uses random samples to approximate a quantity. In the unit square, the quarter-circle has area $\pi/4$. The fraction of random points satisfying $x^2+y^2\leq1$ estimates that area. Multiply by 4 to estimate $\pi$.

**Predict:** out of 2,000 points, will more land inside or outside?

```python
import random
import math
import matplotlib.pyplot as plt

rng = random.Random(2502)
number_of_points = 2000
inside_x, inside_y, outside_x, outside_y = [], [], [], []
for _ in range(number_of_points):
    x, y = rng.random(), rng.random()
    if x**2 + y**2 <= 1:
        inside_x.append(x)
        inside_y.append(y)
    else:
        outside_x.append(x)
        outside_y.append(y)
estimate = 4 * len(inside_x) / number_of_points
print("Estimate:", estimate)
print("Absolute error:", abs(estimate - math.pi))
```

## Run and change
A **seed** fixes the pseudorandom sequence so a rerun is comparable. Change `number_of_points` to 200 and then 10,000, rerunning the whole cell. More points generally improve precision, but the error need not decrease in every individual run. Change the seed to see variability.

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(inside_x, inside_y, s=8, marker="o", label="Inside quarter-circle")
ax.scatter(outside_x, outside_y, s=10, marker="x", label="Outside quarter-circle")
ax.set(xlabel="x", ylabel="y", title="Monte Carlo estimate of pi", aspect="equal")
ax.legend()
plt.show()
```
