## Understand: class vs. object

A **class** is a blueprint for a kind of object. An **instance** is one actual object created from that class. Attributes store data; methods are functions that belong to the object.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def birthday(self):
        self.age += 1

    def __str__(self):
        return f"{self.name}, age {self.age}"


alex = Person("Alex", 19)
print(alex)
```

## What is `self`?

`self` is the instance currently using the method. When we call `alex.birthday()`, Python uses `alex` as `self`. That is how the method knows which object's `age` to change.

## Class attributes vs. instance attributes

An instance attribute belongs to one object. A class attribute is shared as a default by the class.

```python
class Counter:
    category = "simple counter"  # class attribute

    def __init__(self, start=0):
        self.value = start  # instance attribute

    def increment(self):
        self.value += 1


a = Counter()
b = Counter(10)
a.increment()
print(a.value, b.value, Counter.category)
```

## Use: inheritance

Inheritance lets a new class reuse and specialize an existing class.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)


rectangle = Rectangle(3, 5)
square = Square(4)
print(rectangle.area())
print(square.area())
```

## A familiar inheritance example

```python
class Enemy:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def describe(self):
        return f"{self.name}: {self.health} HP"


class Goomba(Enemy):
    def __init__(self):
        super().__init__("Goomba", 1)


print(Goomba().describe())
```

## Why OOP matters even if you rarely write classes

You already use objects constantly. A pandas DataFrame has methods such as `df.head()`. A `requests.Response` object has `response.json()`. A Matplotlib Axes object has `ax.plot()`. The dot means: **ask this object to use one of its methods or attributes**.
