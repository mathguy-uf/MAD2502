## 6. First result

Start with a simple result that directly relates to your question.


````python
group_summary = (
    data.groupby("group", as_index=False)
    .agg(
        number_of_rows=("value", "count"),
        mean_value=("value", "mean"),
        median_value=("value", "median"),
    )
)

group_summary
````

## 7. One useful graph

A graph should answer a question. Always label the axes and give the figure a useful title.


````python
ax = group_summary.plot(
    x="group",
    y="mean_value",
    kind="bar",
    legend=False,
)

ax.set(
    xlabel="Group",
    ylabel="Mean value",
    title="Mean value by group",
)

plt.tight_layout()
plt.show()
````
